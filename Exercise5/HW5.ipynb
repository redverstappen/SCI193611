# HW5: Probability


```python
# โอกาสเป็นคนสูบบุหรี่
s = 0.30

# โอกาสไม่สูบบุหรี่
not_s = 0.70

# โอกาสเป็นมะเร็ง ถ้าเป็นคนสูบบุหรี่
c_if_s = 0.05

# โอกาสเป็นมะเร็ง ถ้าไม่สูบบุหรี่
c_if_not_s = 0.01

# โอกาส X-ray เป็นบวก ถ้าเป็นมะเร็ง
x_if_c = 0.90

# โอกาส X-ray เป็นบวก ถ้าไม่เป็นมะเร็ง
x_if_not_c = 0.20
```

ข้อ 1 P(S=true, C=true, T=บวก)

```python
answer1 = s * c_if_s * x_if_c

print("ข้อ 1")
print("P(S=true, C=true, T=บวก) =", answer1)
print()
```

ข้อ 42 P(T=บวก)


```python
# หา P(C=true)
c = (s * c_if_s) + (not_s * c_if_not_s)

print("P(C=true) =", c)
print()

x = (c * x_if_c) + ((1 - c) * x_if_not_c)

print("ข้อ 42")
print("P(T=บวก) =", x)
print()
```


ข้อ 43 P(C=true | T=บวก)


```python
answer43 = (c * x_if_c) / x

print("ข้อ 43")
print("P(C=true | T=บวก) =", answer43)
print()
```

ข้อ 44 P(S=true | T=บวก)


```python
# โอกาส X-ray เป็นบวก ถ้าเป็นคนสูบบุหรี่
x_if_s = (c_if_s * x_if_c) + ((1 - c_if_s) * x_if_not_c)

# โอกาสเป็นคนสูบบุหรี่และ X-ray เป็นบวก
s_and_x = s * x_if_s

# โอกาสเป็นคนสูบบุหรี่ เมื่อรู้ว่า X-ray เป็นบวก
answer44 = s_and_x / x

print("ข้อ 44")
print("P(S=true | T=บวก) =", answer44)
print()
```

 ข้อ 45 จากคน 1,000 คน จะมีคนที่เป็นมะเร็งและ X-ray เป็นบวกประมาณกี่คน


```python
people = 1000

answer45 = people * c * x_if_c

print("ข้อ 45")
print("จำนวนคนที่เป็นมะเร็งและ X-ray เป็นบวก =", answer45)
print("ประมาณ", round(answer45), "คน")
```
